```python
#sort list
fruits=["apple", "banana", "cherry"]
fruits.sort() #sort in ascending order
print(fruits)
fruits.sort(reverse=True) #sort in descending order
print(fruits)
```

```python
#WAP to check whether the number is divisible by 7 or not
n=int(input("Enter the number to be divided"))
div=7
rem=n%div
if n%div==0:
    print(f"Since remainder is {rem} So,{n} is divisible {div}")
else:
    print(f"Since remainder is {rem} So,{n} is not divisible by {div}")
```

```python
"""WAP to calculate the electricy bill (Accept number of units from user) according to following criteria
Unit               Price
First 100          0
Next 100           Rs 5/unit
After 200          Rs 10/Unit"""

unit=float(input("Enter the Unit of Bill"))
if unit>200:
    print("Your bill amount is = ",500+(unit-200)*10)
elif unit>100 and unit<=200:
    print("Your bill amount is = ",unit*5)
else:
    print("Your Bill amount is Zero")
```
